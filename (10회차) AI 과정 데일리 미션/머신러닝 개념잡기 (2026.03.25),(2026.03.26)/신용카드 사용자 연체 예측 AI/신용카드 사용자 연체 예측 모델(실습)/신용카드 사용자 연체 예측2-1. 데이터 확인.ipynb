### **2. describe()**

📜 **기초 통계량**<br>
* count: 데이터 개수
* mean: 평균
* std: 표준편차
* min: 최솟값
* 25%: 제1 사분위수 (Q1)
* 50%: 제2 사분위수, 중간값 (Q2)
* 75%: 제3 사분위수 (Q3)
* max: 제4 사분위수 (Q4)
* unique (수치형 X): 서로 다른 값의 개수
* top (수치형 X): 최빈값
* freq (수치형 X): 최빈값의 빈도

`describe()`는 주로 수치형 데이터를 간략하게 파악하고 싶을 때 유용하게 사용할 수 있습니다.<br>

```python
# train 데이터의 기초 통계량을 간략하게 확인합니다.
train.describe()
```

* DAYS_BIRTH: 출생일
							데이터 수집 당시 (0)부터 역으로 셈, 즉, -1은 데이터 수집일 하루 전에 태어났음을 의미

* DAYS_EMPLOYED: 업무 시작일
							데이터 수집 당시 (0)부터 역으로 셈, 즉, -1은 데이터 수집일 하루 전부터 일을 시작함을 의미

							양수 값은 고용되지 않은 상태를 의미함
* begin_month: 신용카드 발급 월
							데이터 수집 당시 (0)부터 역으로 셈, 즉, -1은 데이터 수집일 한 달 전에 신용카드를 발급함을 의미

```python
# train 각 변수들을 날짜 단위가 아닌 연 단위(만 나이)로 바꿔보세요 (begin_month 제외)

train.loc[train['DAYS_EMPLOYED'] > 0, train['DAYS_EMPLOYED']] = 0
train['DAYS_BIRTH'] = (train['DAYS_BIRTH'] * -1) / 365.0
train['DAYS_EMPLOYED'] = (train['DAYS_EMPLOYED'] * -1 ) / 365.0
train['begin_month'] = (train['begin_month'] * -1)
```

```python
# test 각 변수들을 날짜 단위가 아닌 연 단위(만 나이)로 바꿔보세요 (begin_month 제외)

test.loc[test['DAYS_EMPLOYED'] > 0, test['DAYS_EMPLOYED']] = 0
test['DAYS_BIRTH'] = (test['DAYS_BIRTH'] * -1) / 365.0
test['DAYS_EMPLOYED'] = (test['DAYS_EMPLOYED'] * -1 ) / 365.0
test['begin_month'] = (test['begin_month'] * -1)
```

### **3. value_counts()**

```python
# train 데이터의 카테고리 컬럼들을 추출합니다.
categories = train.select_dtypes(exclude='number')

# 추출한 변수들의 값들의 비율을 확인합니다.
for feature in categories:
  print(train[feature].value_counts(normalize=True))
  print('-'*50)
```

## **Step 2. 결측치 처리하기**

```python
# train 데이터의 `occyp_type` 변수의 결측치들을 "No job"으로 채워주세요
train["occyp_type"] = train["occyp_type"].fillna("No job")
```

```python
# test 데이터의 `occyp_type` 변수의 결측치들을 "No job"으로 채워주세요
test["occyp_type"] = test["occyp_type"].fillna("No job")
```

```python
# occyp_type 변수의 값의 비율을 백분율로 확인합니다.
round(train["occyp_type"].value_counts(normalize=True) * 100, 2)
```

## **Step 3. 전처리 데이터 저장하기**

```python
# 전처리된 train, test 데이터를 각각 저장해줍니다.

train.to_csv("train.csv", index=False)
test.to_csv("test.csv", index=False)
```
