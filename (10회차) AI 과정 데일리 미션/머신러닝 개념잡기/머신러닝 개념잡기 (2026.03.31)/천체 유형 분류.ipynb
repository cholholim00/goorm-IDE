# **1.필요한 패키지를 불러오기**


```python
import pandas as pd
import numpy as np
import lightgbm as lgbm
```

# **2. 데이터를 불러옵니다.**

```python
train = pd.read_csv('dataset/train.csv', index_col=0)
test = pd.read_csv('dataset/test.csv', index_col=0)
sample_submission = pd.read_csv('dataset/sample_submission.csv', index_col=0)

column_number = {}
for i, column in enumerate(sample_submission.columns):
    column_number[column] = i

def to_number(x, dic):
    return dic[x]

train['type_num'] = train['type'].apply(lambda x: to_number(x, column_number))

train_x = train.drop(columns=['type', 'type_num'], axis=1)
train_y = train['type_num']
test_x = test
```

#**3. 모델을 훈련합니다.**

```python
model = lgbm.LGBMClassifier(n_estimators=1250, num_leaves=128,learning_rate=0.009)
model.fit(train_x,train_y)
```

# **4. 생성된 모델로 예측 데이터를 생성합니다.**

```python
y_pred = model.predict_proba(test_x)
```

# **5. submission 파일을 생성합니다.**

```python
submission = pd.DataFrame(data=y_pred, columns=sample_submission.columns, index=sample_submission.index)
submission.to_csv('dataset/Baseline.csv', index=True)
```
